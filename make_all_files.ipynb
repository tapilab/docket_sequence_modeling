```python
#get the current directory of the current ipynb
import os
current_dir = os.getcwd()
print(current_dir)
```

```python
import shutil
from pathlib import Path

# Get all subdirectories in the current directory
subdirs = [d for d in Path(current_dir).iterdir() if d.is_dir() and not d.name.startswith('.')]

print(f"Found {len(subdirs)} subdirectories:")
for subdir in subdirs:
    print(f"  - {subdir.name}")

# Move all files from subdirectories to current directory
moved_files = []
for subdir in subdirs:
    for file in subdir.rglob('*'):
        if file.is_file():
            # Create new filename to avoid conflicts
            new_name = file.name
            new_path = Path(current_dir) / new_name
            
            # Handle name conflicts by adding the subdirectory name as prefix
            counter = 1
            while new_path.exists():
                stem = file.stem
                suffix = file.suffix
                new_name = f"{subdir.name}_{stem}_{counter}{suffix}"
                new_path = Path(current_dir) / new_name
                counter += 1
            
            # Move the file
            shutil.move(str(file), str(new_path))
            moved_files.append((file, new_path))
            print(f"Moved: {file.relative_to(current_dir)} -> {new_name}")

print(f"\nMoved {len(moved_files)} files")

# Delete the now-empty subdirectories
for subdir in subdirs:
    shutil.rmtree(subdir)
    print(f"Deleted directory: {subdir.name}")

print("\nDirectory flattening complete!")
```
