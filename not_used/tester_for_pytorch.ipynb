```python
import captum
import captum.attr as attr
import sys
import os

print("python:", sys.executable)

print("captum module:", captum)
print("captum.__file__:", getattr(captum, "__file__", None))
print("captum.__path__:", getattr(captum, "__path__", None))

print("attr module:", attr)
print("attr.__file__:", getattr(attr, "__file__", None))

print("has IntegratedGradients:", hasattr(attr, "IntegratedGradients"))
print("IG-related names:", [x for x in dir(attr) if "Integrated" in x or "Layer" in x][:20])

print("shadowing in cwd:", [x for x in os.listdir(".") if x.lower().startswith("captum")])
```
