```python
from pathlib import Path

def get_current_dir():
    try:
        return Path(__file__).resolve().parent
    except NameError:
        # Jupyter / interactive
        return Path.cwd()

CURRENT_DIR = get_current_dir()
```

```python
#remove all init files even in subfolders
for file in CURRENT_DIR.rglob('init.py'):
    file.unlink()
```

```python
#get rid of all of the .ipynb_checkpoints and pycache__ folders even the ones in subfolders and delete the folders themselves as well
for folder in CURRENT_DIR.rglob('__pycache__'):
    for file in folder.iterdir():
        file.unlink()
    folder.rmdir()
for folder in CURRENT_DIR.rglob('.ipynb_checkpoints'):
    for file in folder.iterdir():
        file.unlink()
    folder.rmdir()
```

```python
#get rid of the files and folders within not_used folder and remove the not_used folder itself
not_used_folder = CURRENT_DIR / 'not_used'
for file in not_used_folder.rglob('*'):
    if file.is_file():
        file.unlink()
not_used_folder.rmdir()
```
